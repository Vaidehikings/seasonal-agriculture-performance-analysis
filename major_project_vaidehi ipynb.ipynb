## 6. Data Cleaning
Rainfall, Soil Moisture and Yield had some missing values. Instead of filling them with one overall average, I filled them using the median of their respective Season (and Crop, for Yield) group, since these values naturally change depending on season and crop.

```python
# Clean missing values (fill using season/crop medians, not one overall average)
df['Rainfall_mm'] = df.groupby('Season')['Rainfall_mm'].transform(lambda x: x.fillna(x.median()))
df['Soil_Moisture_pct'] = df.groupby('Season')['Soil_Moisture_pct'].transform(lambda x: x.fillna(x.median()))
df['Yield_Tonnes_Ha'] = df.groupby(['Season','Crop'])['Yield_Tonnes_Ha'].transform(lambda x: x.fillna(x.median()))

df.isnull().sum()
```

### **Key Insights**

Kharif performs the best overall, with the highest yield (5.63 t/ha) and profit (Rs 1,78,915). Zaid is the weakest, with the lowest yield (4.64 t/ha) and a negative average profit, meaning it mostly runs at a loss.

Kharif gets the most rainfall and humidity, Zaid is the hottest and driest. Flood irrigation is the most used method in all seasons despite being less efficient than Drip.

Every crop follows the same order, Kharif highest, then Rabi, then Zaid lowest. Sugarcane gives by far the highest yield.

Revenue drops from Kharif to Zaid but cost stays almost the same, which is why Zaid ends up in loss.

The pattern is not consistent everywhere, Punjab does best in Rabi and Karnataka does best in Zaid.

Rainfall, temperature and humidity all show very low correlation with yield, so other factors like soil or irrigation likely matter more.

### **Recommendations**

Farmers should be cautious about Zaid season farming since profit is negative on average.

Switching from Flood to Drip irrigation, especially in Zaid season, could improve water efficiency.

Punjab and Karnataka should plan around their own best season (Rabi and Zaid respectively) instead of following the general Kharif trend.

Since weather has low correlation with yield, checking soil nutrients and seed quality against yield could give better insights.

### **Conclusion**

This project analyzed seasonal agricultural performance using 4000 farm records across Kharif, Rabi and Zaid seasons. Kharif consistently gives the best yield and profit, while Zaid tends to run at a loss due to lower revenue rather than higher cost. The pattern is not the same for every state, showing that region matters along with season. Weather factors alone did not strongly explain yield differences, suggesting other factors like soil or irrigation play a bigger role. Overall, this suggests agricultural planning should be more region specific rather than one size fits all.
